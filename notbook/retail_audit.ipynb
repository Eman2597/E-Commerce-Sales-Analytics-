# Cleaning customers

```python
import pandas as pd
import numpy as np

customers = pd.read_csv("../data/raw/customers.csv")

customers.info()
customers.tail()
customers.describe()

# Handling data types and duplicates
customers = customers.drop_duplicates(subset=['CustomerID'], keep='first')

customers['Signup_Date'] = pd.to_datetime(customers['Signup_Date'], errors='coerce')

customers['Region'] = (customers['Region'].str.strip().str.lower().str.title())

# check data after cleaning
customers.isna().sum() # noting missing values
customers['Region'].value_counts()
customers.duplicated('CustomerID').sum()
customers.describe(include='all')
customers.head(100)

# Save the cleaned data 
customers.to_csv("../data/processed/customers_clean.csv", index=False)
```

# Cleaning products

```python
products = pd.read_csv("../data/raw/products.csv")

products.info()
products.head()
products.describe(include='all')

# Handling data types and duplicates
products = products.drop_duplicates(subset=['ProductID'], keep='first')
products['Price'] = pd.to_numeric(products['Price'], errors='coerce')
products['Category'] = (products['Category'].str.strip().str.lower().str.title())

# check data after cleaning
products.isna().sum() # noting missing values
products['Category'].value_counts()
products.duplicated('ProductID').sum()
products.describe(include='all')
products.head(100)

# Save the cleaned data
products.to_csv("../data/processed/products_clean.csv", index=False)

```

# Cleaning sales_transactions

```python
sales = pd.read_csv("../data/raw/sales_transactions.csv")

sales.info()
sales.head()
sales.describe(include='all')
# Handling data types and duplicates
sales = sales.drop_duplicates(subset=['TransactionID'], keep='first')
sales['Date'] = pd.to_datetime(sales['Date'], errors='coerce')
sales['Quantity'] = pd.to_numeric(sales['Quantity'], errors='coerce')   
sales['Discount'] = pd.to_numeric(sales['Discount'], errors='coerce')
sales['Total_Amount'] = (
    sales['Total_Amount']
    .astype(str)
    .str.replace(r'[^0-9.-]', '', regex=True))

sales['Total_Amount'] = pd.to_numeric( sales['Total_Amount'], errors='coerce')

sales['ProductID']= sales['ProductID'].astype(str).str.strip()
sales['CustomerID']= sales['CustomerID'].astype(str).str.strip()

sales['Date'] = sales['Date'].fillna(value=pd.NaT).ffill()


# check data after cleaning
sales.isna().sum() # noting missing values

sales.info()
sales.describe(include='all')

# Save the cleaned data
sales.to_csv("../data/processed/sales_transactions_clean.csv", index=False)
```
